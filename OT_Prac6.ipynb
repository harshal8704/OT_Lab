Aim : To find the optimum value of a nonlinear function using Newton’s Optimization Algorithm.

```python
import numpy as np
```

```python
# Function
def f(x):
    return x**2 - 4*x + 4
```

```python
# First derivative
def df(x):
  return 2*x - 4
```

```python
# Second derivative
def d2f(x):
    return 2
```

```python
# Initial guess
x = 0
```

```python
# Tolerance
tol = 1e-6
```

```python
# Maximum iterations
max_iter = 20
```

```python
print("Iteration\t x")
```

```python
for i in range(max_iter):

    x_new = x - df(x)/d2f(x)

    print(i+1, "\t\t", round(x_new,6))

    if abs(x_new - x) < tol:
        break
    x = x_new
```

```python
print("\nOptimal Point =", x_new)
print("Minimum Function Value =", f(x_new))
```
