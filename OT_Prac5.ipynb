Aim : To solve a constrained optimization problem involving inequality constraints using the Kuhn-Tucker (KKT) Conditions.

```python
import sympy as sp
```

```python
# Define variables
x, lam = sp.symbols('x lam', nonnegative=True)
```

```python
# Objective function
f = x**2
```

```python
# Constraint: x >= 2
g = 2 - x
```

```python
# Lagrangian
L = f + lam * g
```

```python
# Stationarity condition
eq1 = sp.diff(L, x)
```

```python
# Complementary slackness
eq2 = lam * g
```

```python
# Solve equations
solutions = sp. solve ((eq1, eq2), (x, lam), dict=True)
```

```python
print("Possible Solutions:")
print(solutions)
```

```python
for sol in solutions:
    x_val = sol[x]
    lam_val = sol[lam]

    if x_val >= 2 and lam_val >= 0:
        print("\nOptimal Solution")
        print("x =", x_val)
        print ("Lambda =", lam_val)
        print ("Minimum Value =", f.subs(x, x_val))
```
